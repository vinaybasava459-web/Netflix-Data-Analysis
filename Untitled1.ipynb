```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('netflix_titles.csv')

# Clean the data
df = df.dropna(subset=['type', 'release_year'])

# Make plots look nice
sns.set_theme(style="whitegrid")
print("Data loaded successfully. Total rows:", len(df))


```

```python
#PLOT 1

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='type', palette='Set1')
plt.title('Distribution of Movies vs TV Shows on Netflix')
plt.xlabel('Content Type')
plt.ylabel('Count')
plt.savefig('plot1_bar.png', bbox_inches='tight')
plt.show()
```

```python
#PLOT 2

plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x='release_year', y='show_id', hue='type', alpha=0.5)
plt.title('Release Year vs. Show ID')
plt.xlabel('Release Year')
plt.ylabel('Show ID')
plt.savefig('plot2_scatter.png', bbox_inches='tight')
plt.show()
```

```python
#PLOT 3

plt.figure(figsize=(6, 4))
sns.boxplot(data=df, x='type', y='release_year', palette='Pastel2')
plt.title('Distribution of Release Years by Content Type')
plt.xlabel('Content Type')
plt.ylabel('Release Year')
plt.savefig('plot3_box.png', bbox_inches='tight')
plt.show()
```
